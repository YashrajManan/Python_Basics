```python

name = "YASHRAJ"
name_len = len(name)
rows = 6
i = 0

while i < rows:
    j = 0
    while j <= 4 * rows:
        dist1 = ((i - rows) ** 2 + (j - rows) ** 2) ** 0.5
        dist2 = ((i - rows) ** 2 + (j - 3 * rows) ** 2) ** 0.5
        if (dist1 < rows + 0.5 or dist2 < rows + 0.5):
            print("$", end="")
        else:
            print(" ", end="")
        j += 1
    print()
    i += 1

# Inverted triangle bottom part
i = 1
while i <= 2 * rows:
    j = 0
    while j < i:
        print(" ", end="")
        j += 1

    j = 0
    while j < 4 * rows + 1 - 2 * i:
        # Print name in the center row
        if i == rows and j == (4 * rows + 1 - 2 * i - name_len) // 2:
            print(name, end="")
            j += name_len
        else:
            print("$", end="")
            j += 1
    print()
    i += 1
```
