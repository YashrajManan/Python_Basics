```python
p=int(input("principal:"))
r=int(input("rate of interest:"))
t=int(input("time:"))
n=int(input("no. of times interest applied:"))
```

```python
simple_interest= ((p*r*t)/100)
```

```python
print(simple_interest)
```

```python
compound_interest= p*(1+(r/n))**(n*t)
```

```python
print(compound_interest)
```
