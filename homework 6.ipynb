```python
#question: Find if value 5 exists in any sublist
```

```python
Input= [[1, 2], [5, 6]] 
i=0 
newlist=[]
while i<len(Input):
    j=0 
    while j<len(Input[i]):
       newlist.append(Input[i][j])
       j+=1 
    i+=1 
if newlist.index(5):
    print("True")
else:
    print("False") 
```

```python
Input= [[1, 2], [3, 4], [2]] 
list_new=[]
i=0 
while i<len(Input):
    j=0
    while j<len(Input[i]):
        if sum(Input[i])>5:
            list_new.append(Input[i][j])
        j+=1 
    i+=1 
print(list_new)
```

```python
#question: Transpose a 2x3 matrix (nested list)
```

```python
Input= [[1, 2, 3], [4, 5, 6]] 
list_new=[]
rows=len(Input)
cols=len(Input[0])
i=0
while i<cols:
    row=[]
    j=0
    while j<rows:
        row.append(Input[j][i])
        j+=1 
    list_new.append(row)
    i+=1 
print(list_new)
```

```python
#question: Filter out all odd numbers from sublists
```

```python
Input= [[1, 2], [3, 4]]
newlist=[]
i=0 
while i<len(Input):
    j=0
    while j<len(Input[i]):
        if Input[i][j]%2==0:
            newlist.append([Input[i][j]])
        j+=1 
    i+=1 
print(newlist)
```
