```python
import pandas as pd
import numpy as np

# Create a random 3D numpy array
data = np.random.rand(3, 4, 5)

print(data)

# Convert the numpy array into a pandas DataFrame
df_3d = pd.DataFrame(data.reshape(3, -1), columns=[f'Col{j+1}' for j in range(4 * 5)])

print(df_3d)

# Access the first "slice" of the 3D DataFrame
slice_0 = df_3d.iloc[0]
print(slice_0)

# Select a specific data point
specific_value = df_3d.iat[0, 3]  # First slice, fourth column
print(specific_value)

# Get the entire second slice
slice_1 = df_3d.iloc[1]
print(slice_1)

# Let's add a new layer to our 3D DataFrame
new_data = np.random.rand(1, 4 * 5)  # One new slice with the same structure
df_3d = pd.concat([df_3d, pd.DataFrame(new_data)], ignore_index=True)

print(df_3d)

# Using .loc for faster access
selected_data = df_3d.loc[0:1]
print(selected_data)

# Simulated weather data: [days, cities, attributes (temp, humidity)]
weather_data = np.random.rand(7, 3, 2)  # 7 days, 3 cities, 2 attributes
weather_df = pd.DataFrame(weather_data.reshape(7, -1), columns=['City1_Temp', 'City1_Humidity', 'City2_Temp', 'City2_Humidity', 'City3_Temp', 'City3_Humidity'])

print(weather_df)

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x, y = np.random.rand(2, 100)  
z = x + y
ax.scatter(x, y, z)
plt.show()

# Assume we already have a dataframe with 3D time series data
time_series_data = np.random.rand(10, 2, 3)  # 10 time points, 2 variables, 3 conditions
time_series_df = pd.DataFrame(time_series_data.reshape(10, -1), columns=['Var1_Cond1', 'Var1_Cond2', 'Var1_Cond3', 'Var2_Cond1', 'Var2_Cond2', 'Var2_Cond3'])

print(time_series_df)
```
